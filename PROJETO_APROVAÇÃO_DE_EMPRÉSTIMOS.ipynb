###**Descrição do Problema**

Bancos e cooperativas de crédito precisam diariamente aprovar ou negar as solicitações de crédito de seus clientes. Os dois grandes problemas são: o risco financeiro que essas instituições correm ao conceder empréstimos a clientes possivelmente inadimplentes, e o risco de perder a oportunidade negando o crédito a clientes confiáveis.
Para tornar este processo de decisão mais rápido e preciso, a análise de dados pode ser utilizada para verificar o histórico dos clientes  e identificar os fatores de aprovação de empréstimo.


###**Objetivo**

O principal objetivo deste projeto é realizar uma análise exploratória dos dados, identificando os  fatores importantes para aprovação de empréstimos.

##**Transformação dos Dados**

Nesta etapa, são criadas variáveis derivadas, como faixa etária e nível de renda. O principal objetivo é facilitar análise e a construção do dashboard.

```python
#Faixa etária
bins = [0, 31, 61, df['idade'].max()+1]
labels = ['Jovem (até 30)', 'Adulto (31-60)', 'Idoso (60+)']

df['faixa_etaria'] = pd.cut(df['idade'], bins=bins, labels=labels, right=False)
```

```python
#Nível de renda
bins = [0, 10000, 50000, 100000, df['renda_anual'].max()+1]
labels = ['Baixa', 'Média', 'Alta', 'Muito Alta']

df['nivel_renda'] = pd.cut(df['renda_anual'], bins=bins, labels=labels, right=False)
```

####**Taxa de aprovação por gênero**

```python
#Tabela de aprovação por gênero.
pd.crosstab(df['genero'], df['aprovacao'])
```

```python
#Calculando a taxa de aprovação
percentual_aprovacao = (df.groupby('genero')['aprovacao'].value_counts(normalize=True).rename('percentual').mul(100).reset_index())
```

```python
#Percentual da aprovação por gênero
plt.figure(figsize=(10,8))
ax = sns.barplot(data=percentual_aprovacao, x='genero', y='percentual', hue='aprovacao', palette='Greens')

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3)


plt.title('Percentual da aprovação do empréstimo por gênero')
plt.xlabel('Gênero')
plt.ylabel('Percentual (%)')
plt.ylim(0,110)
plt.show()
```

Observando o gráfico, nota-se que 77,8% das solicitações foram rejeitas, e 22,2% aprovadas, tanto para mulheres quanto para homens.

##**Insights da Análise**

*  **Relação Renda vs. Solicitação:** A maioria dos solicitantes possui uma renda anual baixa, então solicitam empréstimos entre 5.000 e 12.000, por isso a relação empréstimo/renda é concentrada em valores baixos.

*   **Rigor no Histórico de crédito:** A política de concessão de crédito é conservadora quanto ao passado financeiro, visto que 100% dos que possuem alguma inadimplência anterior tiveram o crédito negado. Embora o histórico não seja único filtro analisado pela empresa, ele atua como principal barreira.

*   **Limitações do Score:** O score de crédito (pontuação) não é o fator de maior peso na decisão final, há um conjunto que influencia na aprovação. Compreende-se isso ao verificar que pessoas com score muito altos (780+) tiveram suas solicatações rejeitadas.

*   **Perfil e Finalidade:** O público majoritariamente jovem busca crédito principalmente para educação, o que sugere que crédito está sendo usado como investimento em carreira. Contudo, a empresa não possui nenhuma garantia de que este público será de bons pagadores, então rejeita o pedido de uma boa parcela.
